```python
import cv2
import numpy as np
import face_alignment
from skimage import io
import torch
import matplotlib.pyplot as plt

fa = face_alignment.FaceAlignment(face_alignment.LandmarksType.TWO_D, device='cuda' if torch.cuda.is_available() else 'cpu')
```

```python
# Initialize the face alignment model
path = "dataset_crop_test/2/minhvb_1.jpg"
# Load an image
image = io.imread(path)

# Get the 2D landmarks
landmarks = fa.get_landmarks(image)

# Plot the landmarks
plt.imshow(image)
for point in landmarks[0]:
    plt.scatter(point[0], point[1], s=10, c='red')
plt.show()
```
